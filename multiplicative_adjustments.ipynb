```python
import pandas as pd
```

```python
best_predictions = pd.read_csv("data/xgboost_model2_new_predictions.csv")
```

```python
weight_factors = [0.58]
for weight_factor in weight_factors:
    weighted_predictions = best_predictions.copy()
    weighted_predictions['order_shipped'] = weight_factor * weighted_predictions['order_shipped']
    weighted_predictions.to_csv(f"data/xgboost_model2_old_predictions_weight_{weight_factor}.csv", index=False)
```
