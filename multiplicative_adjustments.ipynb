```python
import pandas as pd
```

```python
best_predictions = pd.read_csv("data/best_predictions.csv")
```

```python
weight_factors = [0.71, 0.73, 0.74, 0.76]
for weight_factor in weight_factors:
    weighted_predictions = best_predictions.copy()
    weighted_predictions['order_shipped'] = weight_factor * weighted_predictions['order_shipped']
    weighted_predictions.to_csv(f"data/weighted_predictions_{weight_factor}.csv", index=False)
```
